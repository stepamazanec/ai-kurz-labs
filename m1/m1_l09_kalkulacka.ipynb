```python
def odhad(parametry_mld, bity, vram_gb=6, rezerva_gb=1.5, bw_vram=336, bw_ram=80):
    velikost = parametry_mld * bity / 8               # GB vah (M1.L01)
    misto_ve_vram = vram_gb - rezerva_gb               # VRAM pro váhy, bez rezervy
    ve_vram = min(velikost, misto_ve_vram)             # kolik vah se vejde na GPU
    v_ram = velikost - ve_vram                         # zbytek jde do RAM
    cas = ve_vram / bw_vram + v_ram / bw_ram           # sčítají se ČASY (M1.L08)
    print(f"{parametry_mld}B, {bity} bitů: {velikost:.1f} GB | "
          f"VRAM {ve_vram:.1f} GB, RAM {v_ram:.1f} GB | "
          f"max ≈ {1/cas:.0f} tokenů/s")

odhad(7, 16)
odhad(7, 4)
odhad(3, 4)
odhad(14, 4)

# Srovnání formátů pro 7B model (M1.L03)
for b in (16, 8, 4):
    odhad(7, b)
```
